## 5. Sự khác biệt giữa các thước đo Q1, Q2, Q3 trong biểu đồ hộp (Boxplot)

### Cấu trúc của Boxplot:
- **Hộp (Box):** kéo dài từ **Q1 đến Q3**, được gọi là **IQR (Interquartile Range)**  
  `IQR = Q3 – Q1`  
- **Đường ở giữa hộp:** là **Q2 (Median)**.  
- **Whiskers (râu biểu đồ):** mở rộng từ hộp để biểu diễn phạm vi giá trị “bình thường” của dữ liệu.  
  Các điểm nằm ngoài whiskers được xem là **ngoại lai (outliers)**.

### Ý nghĩa các tứ phân vị:
| Ký hiệu | Tên gọi | Mô tả & Ý nghĩa |
|----------|----------|----------------|
| **Q1** (First Quartile – Tứ phân vị thứ nhất) | Giá trị tại **25% dữ liệu nhỏ nhất** | Ranh giới nhóm dữ liệu thấp. |
| **Q2** (Second Quartile – Median – Trung vị) | Giá trị chia dữ liệu thành **hai nửa bằng nhau (50%)** | Ít bị ảnh hưởng bởi ngoại lai, thể hiện trung tâm của dữ liệu. |
| **Q3** (Third Quartile – Tứ phân vị thứ ba) | Giá trị tại **75% dữ liệu nhỏ nhất** | Cho biết ngưỡng mà 75% dữ liệu nhỏ hơn hoặc bằng. |

**Tóm lại:**  Boxplot giúp nhanh chóng nhìn thấy **phân phối dữ liệu, độ phân tán, trung vị và ngoại lai** chỉ trong một biểu đồ.


## 6. Cách xử lý giá trị thiếu (Missing Values) trước khi tính toán thống kê mô tả

### Các phương pháp phổ biến:

| Phương pháp | Cú pháp trong Pandas | Khi nào sử dụng |
|--------------|----------------------|----------------|
| **Loại bỏ giá trị thiếu (Drop)** | `df.dropna()` | Khi dữ liệu thiếu ít, không ảnh hưởng đến mẫu. |
| **Thay thế bằng trung bình (Mean Imputation)** | `df.fillna(df.mean())` | Khi dữ liệu **phân phối chuẩn (normal)**. |
| **Thay thế bằng trung vị (Median Imputation)** | `df.fillna(df.median())` | Khi dữ liệu có **nhiều ngoại lai (outliers)**. |
| **Thay thế bằng mode (Mode Imputation)** | `df.fillna(df.mode().iloc[0])` | Khi dữ liệu là **phân loại (categorical)**. |
| **Nội suy (Interpolation)** | `df.interpolate()` | Khi dữ liệu có **trình tự (time series)**. |
| **KNN Imputation / Mô hình dự đoán** | Dựa vào mô hình ML để ước lượng | Khi cần ước lượng giá trị thiếu dựa trên mối quan hệ giữa các biến. |

*Nguyên tắc:*  Trước khi xử lý giá trị thiếu, cần **xác định nguyên nhân và tỷ lệ thiếu**, tránh làm sai lệch đặc tính của dữ liệu.


## 7. Cách đọc và diễn giải biểu đồ Histogram và Boxplot

### **Histogram (Biểu đồ tần suất)**

| Thành phần | Ý nghĩa |
|-------------|----------|
| **Trục X** | Các khoảng giá trị (bins) |
| **Trục Y** | Tần suất (frequency) hoặc mật độ (density) |
| **Mục đích** | Hiển thị **phân phối của dữ liệu số** |
 **Cách diễn giải:**
- Nếu histogram **đối xứng** → dữ liệu có thể **phân phối chuẩn (normal)**.  
- Nếu **lệch phải** → nhiều giá trị nhỏ, ít giá trị lớn (ví dụ: thu nhập).  
- Nếu **lệch trái** → ngược lại, nhiều giá trị lớn, ít giá trị nhỏ.  
- **Đa đỉnh (multimodal)** → có nhiều nhóm hoặc phân cụm trong dữ liệu.  
 *Ví dụ:*  Histogram thu nhập thường **lệch phải** vì phần lớn người có thu nhập trung bình, chỉ một nhóm nhỏ có thu nhập rất cao.

### **Boxplot (Biểu đồ hộp)**

| Thành phần | Ý nghĩa |
|-------------|----------|
| **Hộp (Box)** | Từ **Q1 đến Q3**, thể hiện độ phân tán dữ liệu (IQR). |
| **Đường giữa hộp** | Trung vị (Q2). |
| **Whiskers** | Mức giá trị bình thường của dữ liệu (thường là ±1.5×IQR). |
| **Điểm ngoài Whiskers** | Các **ngoại lai (outliers)**. |

**Cách diễn giải:**
- Hộp dài → dữ liệu **phân tán lớn**.  
- Trung vị lệch về một phía → dữ liệu **bị lệch (skewed)**.  
- Nhiều điểm ngoài → dữ liệu có **nhiều ngoại lai**.  

Boxplot rất hữu ích để **so sánh phân phối giữa các nhóm** và **phát hiện dữ liệu bất thường**.

## 8. Cách xử lý giá trị ngoại lai (Outliers) trước khi thực hiện thống kê mô tả

### Các phương pháp xử lý:

| Phương pháp | Mô tả | Khi nào dùng |
|--------------|--------|--------------|
| **Giữ nguyên (Keep)** | Giữ ngoại lai nếu phản ánh thực tế (ví dụ: thu nhập cực cao). | Khi giá trị đó có ý nghĩa thật trong bối cảnh nghiên cứu. |
| **Loại bỏ (Trimming)** | Loại các giá trị ngoài khoảng `[Q1 - 1.5×IQR, Q3 + 1.5×IQR]`. | Khi dữ liệu ngoại lai rõ ràng là lỗi đo lường. |
| **Biến đổi dữ liệu (Transformation)** | Dùng log, sqrt hoặc chuẩn hóa để giảm ảnh hưởng. | Khi vẫn muốn giữ dữ liệu nhưng giảm độ lệch. |
| **Thay thế (Capping/Winsorizing)** | Gán ngoại lai bằng giá trị cận gần nhất (ví dụ: Q1 - 1.5×IQR). | Khi muốn ổn định dữ liệu mà không loại bỏ quan sát. |
 *Nguyên tắc:*  
- **Không nên loại ngoại lai tùy tiện.**  
- Trước khi xử lý, cần **xác định nguyên nhân (lỗi hay giá trị thực)** và **xem xét mục tiêu phân tích**.
