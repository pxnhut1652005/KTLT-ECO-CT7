#### **2.1.1. Ôn tập lý thuyết**

##### *Quy trình khai phá dữ liệu CRISP – DM (Cross Industry Standard Process for Data Mining) là gì ? Quy trình khai phá dữ liệu SEMMA (Sample, Explore, Modify, Model, Access) là gì?*


* **Khái niệm**

  **CRISP–DM** (viết tắt của Cross-Industry Standard Process for Data Mining – Quy trình Tiêu chuẩn Liên ngành cho Khai thác Dữ liệu) là một mô hình quy trình chuẩn công nghiệp được phát triển vào năm 1996 bởi sự hợp tác của Daimler-Benz, SPSS và NCR trong khuôn khổ CRISP–DM Consortium.
- Mục tiêu của CRISP–DM là cung cấp một khung làm việc toàn diện, có thể áp dụng cho mọi ngành nghề, nhằm hướng dẫn các nhà phân tích dữ liệu và tổ chức thực hiện các dự án khai phá dữ liệu một cách có hệ thống, nhất quán và hiệu quả.

* **6 giai đoạn của CRISP - DM**

**Giai đoạn 1: Hiểu biết về kinh doanh (Business Understanding)**
* Mục tiêu chính: Đảm bảo các nỗ lực dựa trên dữ liệu được phù hợp với mục tiêu chiến lược của tổ chức, tránh lãng phí nguồn lực và xác định rõ ràng mục tiêu cùng tiêu chí thành công cho toàn bộ dự án.
* Hoạt động chính: 
  + Cụ thể hóa vấn đề/ kết quả mong muốn
  + Chuyển mục tiêu kinh doanh thành mục tiêu phân tích có thể đo lường được
  + Kiểm kê và đánh giá nguồn lực sẵn có: dữ liệu, công cụ, nền tảng, nhân sự và ngân sách.
  + Xác định hạn chế và rủi ro:  giới hạn thời gian, lo ngại về quyền riêng tư dữ liệu, yêu cầu pháp lý, hạn chế kỹ thuật và cân nhắc về mặt đạo đức.
  + Thiết lập tiêu chí thành công và phạm vi
  + Lập kế hoạch bước đầu

**Giai đoạn 2: Hiểu dữ liệu (Data Understanding)**
* Mục tiêu chính: Thu thập, khám phá và đánh giá dữ liệu thô để hiểu rõ đặc điểm, điểm mạnh và hạn chế của dữ liệu, đảm bảo dữ liệu phù hợp với các nhiệm vụ phân tích hoặc mô hình hóa, đồng thời xây dựng nền tảng vững chắc cho dự án.
* Hoạt động chính:
  + Xác định và truy cập nguồn dữ liệu: thu thập dữ liệu từ các nguồn nội bộ và bên ngoài, đảm bảo đầy đủ và cập nhật.
  + Phân tích siêu dữ liệu/ từ điểm dữ liệu: mục đích để hiểu rõ ý nghĩa, kiểu dữ liệu, phạm vi giá trị và mối quan hệ giữa các biến.
  + Khám phá dữ liệu bằng trực quan hóa: Sử dụng biểu đồ hộp, biểu đồ phân tán, biểu đồ tần suất… để nhận diện xu hướng, mô hình và phân bố dữ liệu.
  + Phân tích thống kê sơ bộ: Áp dụng phân tích tương quan, kiểm định giả thuyết hoặc thống kê mô tả để phát hiện mối quan hệ hoặc các vấn đề tiềm ẩn.
  + Xác định điểm bất thường / ngoại lệ: Phát hiện và ghi chú các giá trị bất thường, outlier, thiếu dữ liệu hoặc sai sót để xử lý trong bước chuẩn bị dữ liệu.

**Giai đoạn 3: Chuẩn bị dữ liệu**
* Mục tiêu chính: Chuẩn bị dữ liệu bằng cách làm sạch, chuyển đổi và định dạng lại dữ liệu thô thành dữ liệu sạch, nhất quán và sẵn sàng cho phân tích hoặc mô hình hóa, nhằm đảm bảo tính chính xác và độ tin cậy của kết quả phân tích.
* Hoạt động chính:
  + Lựa chọn dữ liệu phù hợp với mục tiêu phân tích/mô hình hóa.
  + Loại bỏ dữ liệu không liên quan hoặc dư thừa.
  + Xử lý giá trị thiếu (bằng trung bình, trung vị, mốt, hoặc kỹ thuật khác).
  + Phát hiện và xử lý giá trị ngoại lệ.
  + Chuẩn hóa và thống nhất định dạng dữ liệu, đơn vị, nhãn.
  + Tạo đặc trưng mới thông qua chuyển đổi, tổng hợp hoặc kết hợp.
  + Chuẩn hóa/điều chỉnh tỷ lệ các biến để cân bằng thang đo.
  + Mã hóa biến phân loại (one-hot, label encoding,…).
  + Áp dụng kỹ thuật giảm chiều để tối ưu số lượng biến đầu vào.

**Giai đoạn 4: Mô hình hóa - Xây dựng khung dự đoán hoặc khung hiểu biết**
* Mục tiêu chính: Chuyển dữ liệu đã chuẩn bị thành mô hình hoặc khuôn khổ phân tích để tạo ra dự đoán chính xác hoặc hiểu biết có ý nghĩa phục vụ mục tiêu kinh doanh.
* Hoạt động chính:
  + Chọn kỹ thuật mô hình hóa
  + Tạo thiết kế thử nghiệm
  + Xây dựng mô hình
  + Đánh giá mô hình

**Giai đoạn 5: Đánh giá - Đảm bảo phù hợp với mục tiêu kinh doanh**
* Mục tiêu chính: Đánh giá hiệu suất mô hình và giá trị thực tiễn của các kết quả phân tích, đảm bảo chúng phù hợp với mục tiêu kinh doanh và mang lại tác động rõ ràng.
* Hoạt động chính:
  + Đánh giá toàn diện kết quả
  + Xác định mức độ đáp ứng câu hỏi kinh doanh và tiêu chí thành công.
  + Định lượng tác động lên KPI hoặc chỉ số kinh doanh chính.
  + Trực quan hóa kết quả để dễ hiểu và truyền đạt.
  + Xác định các bước tiếp theo

**Giai đoạn 6: Triển khai - Biến hiểu biết thành kết quả thực tế**
* Mục tiêu chính: Đưa mô hình hoặc hiểu biết phân tích vào thực tế vận hành, đảm bảo chúng tạo ra giá trị và tác động bền vững cho tổ chức.
* Hoạt động chính:
  + Kế hoạch triển khai
  + Kế hoạch giám sát và bảo trì
  + Tạo báo cáo cuối cùng
  + Đánh giá dự án

```python
import numpy as np #numerical computation
import pandas as pd #data wrangling
import matplotlib.pyplot as plt #plotting package
#Next line helps with rendering plots
%matplotlib inline
import matplotlib as mpl #add'l plotting functionality
mpl.rcParams['figure.dpi'] = 400 #high res figures
import graphviz #to visualize decision trees
```

**2. Nạp dữ liệu vào bộ nhớ, phân tích và loại bỏ những features không liên quan đến bài toán cần giải
quyết**

```python
df = pd.read_csv('datasets/default_of_credit_card_clients.csv') #Load the cleaned data
features_response = df.columns.tolist() #Get a list of column names
#Make a list of columns to remove that aren't features or the response variable
items_to_remove = ['ID', 'SEX', 'PAY_2', 'PAY_3',\
 'PAY_4', 'PAY_5', 'PAY_6',\
 'EDUCATION_CAT', 'graduate school',\
 'high school', 'none',\
 'others', 'university']
features_response = [item for item in features_response if item not
in items_to_remove]
features_response
```

**3. Chuẩn bị dữ liệu cho tập train và tập test**

```python
from sklearn.model_selection import train_test_split
from sklearn import tree
#Split the data into training and testing sets using the same random seed
X_train, X_test, y_train, y_test = \
 train_test_split(df[features_response[:-1]].values,
 df['default payment next month'].values,
 test_size=0.2, random_state=24)
```

**4. Xây dựng cây quyết định từ lớp DecisionTreeClassifier có trong thư viện Scikit-Learn**

```python
# the tree will grow to a depth of at most 2
dt = tree.DecisionTreeClassifier(max_depth=2)
dt.fit(X_train, y_train)
```

**5. Hiển thị cây quyết định với package graphviz**

```python
!sudo apt-get update
!sudo apt-get install graphviz -y
!pip install graphviz
```

```python
from sklearn import tree
import graphviz

dot_data = tree.export_graphviz(
    dt,
    out_file=None,
    filled=True,
    rounded=True,
    feature_names=features_response[:-1],
    proportion=True,
    class_names=['Not defaulted', 'Defaulted']
)

graph = graphviz.Source(dot_data)
graph
```
