### **2.1. GIẢI THUẬT 1: CÂY QUYẾT ĐỊNH VÀ RỪNG CÂY**

#### **2.1.1. Ôn tập lý thuyết**

##### *Quy trình khai phá dữ liệu CRISP – DM (Cross Industry Standard Process for Data Mining) là gì ? Quy trình khai phá dữ liệu SEMMA (Sample, Explore, Modify, Model, Access) là gì?*


**CRISP–DM** (viết tắt của Cross-Industry Standard Process for Data Mining – Quy trình Tiêu chuẩn Liên ngành cho Khai thác Dữ liệu)
là một mô hình quy trình chuẩn công nghiệp được phát triển vào năm 1996 bởi sự hợp tác của Daimler-Benz, SPSS và NCR trong khuôn khổ CRISP–DM Consortium.
