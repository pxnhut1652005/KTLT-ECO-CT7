## **THỰC HÀNH: CÁC GIẢI THUẬT PHÂN LOẠI CƠ BẢN**

### **2.3. GIẢI THUẬT 3: BAYES NGÂY THƠ (NAÏVE BAYES)**

#### **2.3.1. Ôn tập lý thuyết**

##### ***Giải thuật Naive Bayes hoạt động như thế nào? Hãy giải thích định lý Bayes và giả định "ngây thơ" trong thuật toán này?***

Naive Bayes là một thuật toán học có giám sát dùng để giải quyết cho các bài toán phân loại, hoạt động dựa trên Định lý Bayes. Đây là một bộ phân loại xác suất, trong đó thuật toán học cách ước lượng xác suất xuất hiện của từng đặc điểm trong mỗi lớp. Khi dự đoán, thuật toán sẽ kết hợp các xác suất này và gán đối tượng vào lớp có xác suất cao nhất.

Định lý Bayes mô tả cách tính xác suất của một giả thuyết dựa trên bằng chứng quan sát được. Nó kết hợp giữa xác suất tiên nghiệm của giả thuyết P(H) và xác suất quan sát bằng chứng khi giả thuyết đúng P(E|H) để suy ra xác suất hậu nghiệm của giả thuyết khi đã có bằng chứng P(H|E). Công thức được viết là:

                                        P(H|E) = (P(E|H) * P(H))/P(E)
- P(H) là xác suất tiên nghiệm của giả thuyết H; 
- P(E|H) là xác suất xảy ra bằng chứng E khi H đúng;
- P(E) là xác suất xảy ra bằng chứng E; 
- P(H|E) là xác suất hậu nghiệm cần tìm.
  
Trong thuật toán Naive Bayes, công thức này được sử dụng để tính xác suất một đối tượng thuộc về từng lớp dựa trên các đặc điểm quan sát, sau đó chọn lớp có xác suất cao nhất.

Giả định “ngây thơ” của thuật toán là các đặc điểm dữ liệu là độc lập có điều kiện khi biết lớp. Nhờ vậy, ta có thể viết P(E|H) dưới dạng tích của các xác suất thành phần, giúp việc tính toán đơn giản hơn nhiều so với việc ước lượng toàn bộ phân phối phức tạp của dữ liệu. Mặc dù giả định này hiếm khi đúng hoàn toàn trong thực tế nhưng nó vẫn cho phép Naive Bayes hoạt động hiệu quả và được ứng dụng rộng rãi trong nhiều bài toán phân loại, đặc biệt là xử lý văn bản và lọc thư rác.

##### ***Các loại mô hình Naive Bayes (Gaussian, Multinomial, Bernoulli) khác nhau ra sao? Khi nào nên sử dụngtừng loại?***

**Gaussian Naive Bayes:**
- Gaussian Naive Bayes được sử dụng khi các thuộc tính của dữ liệu là biến liên tục và giả định rằng trong từng lớp, các biến này tuân theo phân phối Gaussian (hay phân phối chuẩn). Với giả định này, mô hình có thể ước lượng xác suất một cách đơn giản và nhanh chóng mà không cần đến các phép tính phức tạp. Đây là một thuật toán đơn giản nhưng hiệu quả, đặc biệt giúp tăng tốc độ huấn luyện và dự đoán. Trong nhiều tình huống, sai số của Gaussian Naive Bayes chỉ lớn gấp đôi so với Naive Bayes tối ưu, do đó nó vẫn được xem là một lựa chọn thực tế và đáng tin cậy cho dữ liệu liên tục.
- Mô hình này thường được áp dụng trong các bài toán có dữ liệu đo lường, dữ liệu y sinh, phân loại tín dụng,… hoặc các dữ liệu dạng số thực như chiều cao, cân nặng, điểm số hay nhiệt độ.

**Multinomial Naive Bayes:**
- Multinomial Naive Bayes được sử dụng khi các thuộc tính của dữ liệu là rời rạc và biểu diễn dưới dạng tần suất hoặc số lần xuất hiện của sự kiện (giá trị nguyên không âm). Mô hình cho phép ước lượng xác suất một đối tượng thuộc về từng lớp dựa trên phân phối tần suất của các đặc trưng.
- Loại mô hình này đặc biệt phù hợp trong các bài toán phân loại văn bản, nơi dữ liệu thường được biểu diễn bằng vector số lần xuất hiện từ (Bag of Words, TF-IDF). Nhờ cách tiếp cận đó, Multinomial Naive Bayes cũng hoạt động rất hiệu quả trong các ứng dụng như phân loại tài liệu, lọc thư rác, phân tích cảm xúc hay phát hiện chủ đề.

**Bernoulli Naive Bayes:**
- Bernoulli Naive Bayes là một thuật toán phù hợp cho dữ liệu có các thuộc tính nhị phân hoặc boolean, trong đó mỗi đặc trưng chỉ nhận một trong hai giá trị như có/không, đúng/sai, được chấp nhận/bị từ chối,… Khác với Multinomial Naive Bayes, mô hình này không quan tâm số lần một đặc trưng xuất hiện mà chỉ xét đến việc nó có xuất hiện hay không.
- Trong thực tế, Bernoulli Naive Bayes đặc biệt hiệu quả với dữ liệu dạng “có/không”, chẳng hạn như việc kiểm tra từ “học” có xuất hiện trong email hay không. Mô hình này thường được áp dụng trong các bài toán phân loại văn bản ngắn, lọc thư rác hoặc những trường hợp mà sự hiện diện của đặc trưng quan trọng hơn số lần xuất hiện.

##### ***Tại sao Naive Bayes được gọi là “ngây thơ”? Giả định về tính độc lập của các đặc trưng ảnh hưởng như thế nào đến hiệu suất của mô hình?***

Naive Bayes được gọi là “ngây thơ” vì mô hình dựa trên những giả định đơn giản nhưng hiếm khi đúng trong thực tế. Cụ thể:
- Các đặc trưng (features) được coi là độc lập có điều kiện với nhau khi đã biết nhãn lớp, tức là sự thay đổi của một đặc trưng không ảnh hưởng đến các đặc trưng khác.
- Các đặc trưng được giả định có mức độ ảnh hưởng ngang nhau đến biến mục tiêu.

Trên thực tế, các đặc trưng dữ liệu thường có mối quan hệ và tương quan lẫn nhau nên giả định này hiếm khi tồn tại. Chính vì vậy, thuật toán mới được gọi là “ngây thơ”. Tuy nhiên, nhờ giả định này, việc tính toán xác suất chung được đơn giản hóa thành tích của các xác suất riêng lẻ, giúp mô hình nhẹ, dễ triển khai và dự đoán kết quả nhanh chóng.

Mặc dù mang tính “ngây thơ”, Naive Bayes vẫn hoạt động hiệu quả trong nhiều tình huống thực tế, đặc biệt trên các tập dữ liệu lớn. Các ứng dụng phổ biến có thể kể đến là lọc thư rác, phân loại văn bản, phân tích cảm xúc và dự đoán sắc thái ngôn ngữ.

Giả định về tính độc lập của các đặc trưng cũng ảnh hưởng trực tiếp đến hiệu suất của Naive Bayes. Nếu trong dữ liệu giả định này tương đối đúng, mô hình có thể dự đoán nhanh chóng và chính xác với độ phức tạp rất thấp. Ngược lại, khi các đặc trưng có mối liên hệ hoặc phụ thuộc lẫn nhau, mô hình có thể bỏ qua những thông tin quan trọng và dẫn đến phân loại sai. Dù vậy, nhờ cấu trúc đơn giản, Naive Bayes vẫn thường cho hiệu quả đáng tin cậy trong nhiều tình huống, đặc biệt khi số chiều dữ liệu lớn hoặc khi dữ liệu huấn luyện hạn chế.
